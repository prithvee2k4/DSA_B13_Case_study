#1)Employee Performance Analyzer.
* Problem Statement
A company maintains employee records containing details such as Employee ID, Name, Department, Salary, and Performance Rating. The HR department wants to analyze these records using Python to generate useful insights.
Write a Python program to perform various operations on the employee data using only core Python concepts such as variables, data types, operators, control statements, lists, tuples, dictionaries, and sets.
* Scenario
ABC Technologies has stored employee information in a list of dictionaries. The HR manager wants to generate reports such as the total number of employees, salary statistics, department information, and employee search.
Your task is to analyze the employee dataset and display the required information.
Dataset
employees = [
    {"id":101, "name":"Alice", "department":"Data", "salary":65000, "rating":4.5},
    {"id":102, "name":"Bob", "department":"AI", "salary":72000, "rating":4.8},
    {"id":103, "name":"Charlie", "department":"Web", "salary":58000, "rating":3.9},
    {"id":104, "name":"David", "department":"Data", "salary":61000, "rating":4.2},
    {"id":105, "name":"Eva", "department":"AI", "salary":75000, "rating":4.9},
    {"id":106, "name":"Frank", "department":"Cloud", "salary":68000, "rating":4.1},
    {"id":107, "name":"Grace", "department":"Web", "salary":59000, "rating":3.8},
    {"id":108, "name":"Helen", "department":"Data", "salary":70000, "rating":4.6}
]

* Tasks
Your program should perform the following operations:
1) Display the total number of employees.
2) Display the average, highest, and lowest salary.
3. Display the names of employees whose salary is greater than ₹65,000 and whose performance rating is 4.5 or above.
4. Create a dictionary with Employee ID as the key and Employee Name as the value.
5. Create a list containing all employee names, sort it alphabetically, reverse it, and display the first three names.
6. Store all Employee IDs in a tuple and display the maximum and minimum Employee ID.
7. Create a set of unique departments and display the department names.
8. Accept an Employee ID from the user and display the employee details if found; otherwise display "Employee Not Found".






```python
employees = [
    {"id": 101, "name": "Alice", "department": "Data", "salary": 65000, "rating": 4.5},
    {"id": 102, "name": "Bob", "department": "AI", "salary": 72000, "rating": 4.8},
    {"id": 103, "name": "Charlie", "department": "Web", "salary": 58000, "rating": 3.9},
    {"id": 104, "name": "David", "department": "Data", "salary": 61000, "rating": 4.2},
    {"id": 105, "name": "Eva", "department": "AI", "salary": 75000, "rating": 4.9},
    {"id": 106, "name": "Frank", "department": "Cloud", "salary": 68000, "rating": 4.1},
    {"id": 107, "name": "Grace", "department": "Web", "salary": 59000, "rating": 3.8},
    {"id": 108, "name": "Helen", "department": "Data", "salary": 70000, "rating": 4.6}
]

# Q1 Display the total number of employees.
total_employees = len(employees)

print("Total Employees:", total_employees)
print()

# Q2 Display the average, highest, and lowest salary.

salaries = [emp["salary"] for emp in employees]

average_salary = sum(salaries) / len(salaries)
highest_salary = max(salaries)
lowest_salary = min(salaries)

print("Average Salary:", average_salary)
print("Highest Salary:", highest_salary)
print("Lowest Salary:", lowest_salary)

print()

# Q3 Display the names of employees whose salary is greater than ₹65,000 and
# whose performance rating is 4.5 or above

print("Employees with Salary > 65000 and Rating >= 4.5")

for emp in employees:
    if emp["salary"] > 65000 and emp["rating"] >= 4.5:
        print(emp["name"])
print()

# Q4 Create a dictionary with Employee ID as the key and Employee Name as the
# value.

employee_dictionary = {}

for emp in employees:
    employee_dictionary[emp["id"]] = emp["name"]

print("Employee Dictionary")
print(employee_dictionary)
print()

# Q5 Create a list containing all employee names, sort it alphabetically,
# reverse it, and display the first three names.

employee_names = []

for emp in employees:
    employee_names.append(emp["name"])

employee_names.sort()
employee_names.reverse()

print("First Three Employee Names")

for i in range(3):
    print(employee_names[i])

print()

# Q6 Store all Employee IDs in a tuple and display the maximum and minimum
# Employee ID.

employee_ids = tuple(emp["id"] for emp in employees)

print("Employee ID Tuple")
print(employee_ids)
print("Maximum Employee ID:", max(employee_ids))
print("Minimum Employee ID:", min(employee_ids))
print()

# Q7 Create a set of unique departments and display the department names.

departments = set(emp["department"] for emp in employees)

print("Departments")
print(departments)
print()

# Q8 Accept an Employee ID from the user and display the employee details if
# found; otherwise display "Employee Not Found".

search_id = int(input("Enter Employee ID: "))

employee_found = 0

for employee in employees:
    if employee["id"] == search_id:
        print()
        print("Employee Found")
        print("ID:", employee["id"])
        print("Name:", employee["name"])
        print("Department:", employee["department"])
        print("Salary:", employee["salary"])
        print("Rating:", employee["rating"])

        employee_found = 1
        break

if employee_found == 0:
    print()
    print("Employee Not Found")
```
