# Parking Fee Calculator


```python
Vehicle_Type=input("Enter the type of vehicle: ")
Hours=int(input("Enter the number of hours parked: "))
Parking_Fee=0
if Vehicle_Type=="Car":
  if Hours<=2:
    Parking_Fee=50
  else:
    Parking_Fee=50+((Hours-2)*20) # Total hours taken -2 (default 2 hours rate=50).
elif Vehicle_Type=="Bike":
  if Hours<=2:
    Parking_Fee=30
  else:
    Parking_Fee=30+((Hours-2)*10)
elif Vehicle_Type=="SUV":
  if Hours<=2:
    Parking_Fee=70
  else:
    Parking_Fee=70+((Hours-2)*30)
else:
  print("Invalid Vehicle Type")
print("Parking Fee:",Parking_Fee)
```

# Movie Theatre Seat Booking Analyzer
A movie theatre has multiple shows. Each show has a fixed number of seats.
For each show, the following information is provided:
● Show name
● Total seats
● Number of booked seats
The theatre wants to analyze the occupancy of each show.
Occupancy Rules
Calculate:
Occupancy Percentage = (Booked Seats / Total Seats) × 100
Classify the show as:
Occupancy
Category
80% or above House Full
60%–79%
High Demand
40%–59%
Below 40%
Your Task
Moderate
Low Demand
Write a Python program that accepts information for N shows and displays:
1. Show name
2. Available seats
3. Occupancy percentage
4. Occupancy category
At the end, display:
● Show with the highest occupancy
● Total number of booked seats
● Total number of available seats
● Number of shows classified as High Demand or House Full
* Input Format
The first line contains:
N
The next N lines contain:
Show_Name Total_Seats Booked_Seats

```python
shows = int(input())

highest_occupancy = 0
highest_show = ""

total_booked = 0
total_available = 0
high_demand = 0

show_info = []

for i in range(shows):

    data = input().split()

    show_name = data[0]
    total_seats = int(data[1])
    booked_seats = int(data[2])

    available_seats = total_seats - booked_seats

    occupancy = (booked_seats / total_seats) * 100

    if occupancy >= 80:
        category = "House Full"
    elif occupancy >= 60:
        category = "High Demand"
    elif occupancy >= 40:
        category = "Moderate"
    else:
        category = "Low Demand"

    show_info.append([show_name, booked_seats, available_seats,
                      occupancy, category])


for x in show_info:

    show_name = x[0]
    booked_seats = x[1]
    available_seats = x[2]
    occupancy = x[3]
    category = x[4]

    print(show_name, "- Available:", available_seats,
          "- Occupancy:", format(occupancy, ".2f") + "%",
          "-", category)

    if occupancy > highest_occupancy:
        highest_occupancy = occupancy
        highest_show = show_name

    if category == "High Demand" or category == "House Full":
        high_demand += 1

    total_booked += booked_seats
    total_available += available_seats


print("Highest Occupancy:", highest_show)
print("Total Booked Seats:", total_booked)
print("Total Available Seats:", total_available)
print("High Demand/House Full Shows:", high_demand)
```

# Employee Leave Management System


```python
class Employee:

    def __init__(self, employee_id, employee_name, department, leave_balance):
        self.employee_id = employee_id
        self.employee_name = employee_name
        self.department = department
        self.leave_balance = leave_balance

    def apply_leave(self, days):

        if days <= 0:
            print("Invalid Leave Request")

        elif days <= self.leave_balance:
            self.leave_balance = self.leave_balance - days
            print("Leave Approved")

        else:
            print("Insufficient Leave Balance")

    def cancel_leave(self, days):

        if days > 0:
            self.leave_balance = self.leave_balance + days
            print("Leave Balance:", self.leave_balance)

        else:
            print("Invalid Cancellation")

    def check_balance(self):
        print("Leave Balance:", self.leave_balance)

    def display_employee(self):
        print("Employee ID:", self.employee_id)
        print("Employee Name:", self.employee_name)
        print("Department:", self.department)
        print("Leave Balance:", self.leave_balance)


# Taking employee details
data = input().split()

employee_id = data[0]
employee_name = data[1]
department = data[2]
leave_balance = int(data[3])

# Creating object
employee = Employee(employee_id, employee_name, department, leave_balance)

# Number of operations
operations = int(input())

for i in range(operations):

    operation = input().split()

    if operation[0] == "APPLY":
        days = int(operation[1])
        employee.apply_leave(days)

    elif operation[0] == "CANCEL":
        days = int(operation[1])
        employee.cancel_leave(days)

    elif operation[0] == "BALANCE":
        employee.check_balance()

    elif operation[0] == "DISPLAY":
        employee.display_employee()
```
